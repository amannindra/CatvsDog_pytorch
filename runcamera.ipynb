```python
# !conda install -y torchvision

import cv2
from PIL import Image

# PyTorch Library
import torch

import torchvision.models as models
import torchvision.transforms as transforms

import matplotlib.pylab as plt
```

```python
model_path = "model.pth"

device = torch.device("cpu")


def load_model(path):
    model = models.resnet18(pretrained=True)
    model.load_state_dict(torch.load(path, map_location=torch.device("cpu")))
    model.eval()
    return model
```

#### Cat vs Dog Image Dectector




```python
def plot_image(image):
    plt.imshow(image, cmap="gray")
    plt.show()


def predict(model, image_path):
    image = Image.open(image_path)
    preprocess = transforms.Compose(
        [
            transforms.Resize(256),
            transforms.CenterCrop(224),
            transforms.ToTensor(),
            transforms.Normalize(
                mean=[0.485, 0.456, 0.406],  # standard ImageNet normalization
                std=[0.229, 0.224, 0.225],
            ),
        ]
    )
    input_tensor = preprocess(image).unsqueeze(0)

    with torch.no_grad():
        output = model(input_tensor)

    predicted_class = output.argmax(dim=1).item()

    if predicted_class == 0:
        return "cat"
    else:
        return "dog"


model = load_model(model_path)
image_path = "cat1.jpg"
prediction = predict(model, image_path)
print(f"Prediction: {prediction}")
```

#### Cat vs Dog Video Dectector



```python
def predictVideo(model, image_path):
    image_from_np = Image.fromarray(image_path)

    preprocess = transforms.Compose(
        [
            transforms.Resize(256),
            transforms.CenterCrop(224),
            transforms.ToTensor(),
            transforms.Normalize(
                mean=[0.485, 0.456, 0.406],  # standard ImageNet normalization
                std=[0.229, 0.224, 0.225],
            ),
        ]
    )
    input_tensor = preprocess(image_from_np).unsqueeze(0)

    with torch.no_grad():
        output = model(input_tensor)

    predicted_class = output.argmax(dim=1).item()

    return predicted_class


cont = True

cap = cv2.VideoCapture(0)
modelVideo = load_model(model_path)
while cont:
    ret, frame = cap.read()
    if not ret:
        break
    frame = cv2.flip(frame, 1)
    rgb_for_model = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    prediction = predictVideo(modelVideo, rgb_for_model)
    # print(type(frame_rgb))

    if prediction == 0:
        prediction = "cat"
    else:
        prediction = "dog"

    cv2.putText(
        frame, prediction, (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2
    )
    cv2.imshow("frame", frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        cont = False

cap.release()
cv2.destroyAllWindows()
```
